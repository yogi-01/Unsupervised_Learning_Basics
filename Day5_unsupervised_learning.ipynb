```python
import pandas as pd
import seaborn as sn
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("CC GENERAL.csv")
df.head()
```

```python
df.describe()
```

```python
df=df.dropna()
```

```python
df.info()
```

```python
df=df.drop_duplicates()
```

```python
df.head()
```

```python
df=df.drop(columns=['CUST_ID'])
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
```

```python
s=StandardScaler()
df_transform=s.fit_transform(df)
```

```python
inertia=[]

for k in range(1,11):
  model=KMeans(n_clusters=k,random_state=42,n_init=10)
  model.fit(df_transform)
  inertia.append(model.inertia_)
```

```python
plt.plot(range(1,11),inertia,marker='o')
plt.xlabel("No of Clusters(k)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
#K means clustering
kmeans=KMeans(n_clusters=4,random_state=42,n_init=10)
kmeans_res=kmeans.fit_predict(df_transform)
print(kmeans_res)
```

```python
#DBSCAN
dbscan=DBSCAN(eps=2.0,min_samples=10)
dbscan_res=dbscan.fit_predict(df_transform)
```

```python
# PCA - Reduce to 2 dimensions
pca = PCA(n_components=2)
X_pca = pca.fit_transform(df_transform)

print("Explained Variance:", pca.explained_variance_ratio_)
```

```python
# 7. Visualize K-Means
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans_res, cmap="viridis")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("K-Means Clustering")
plt.colorbar(label="Cluster")
plt.show()
```

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=dbscan_res, cmap="viridis")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("DBSCAN Clustering")
plt.colorbar(label="Cluster (-1 = Noise)")
plt.show()
```

```python
# K-Means Cluster Counts
print("K-Means Clusters:")
print(pd.Series(kmeans_res).value_counts().sort_index())

# DBSCAN Cluster Counts
print("\nDBSCAN Clusters:")
print(pd.Series(dbscan_res).value_counts().sort_index())

# DBSCAN Cluster and Noise Summary
n_clusters = len(set(dbscan_res) - {-1})
n_noise = sum(dbscan_res == -1)

print("\nNumber of DBSCAN Clusters:", n_clusters)
print("Number of Noise Points:", n_noise)
```

```python
# K-Means Cluster Profile
df_clusters = df.copy()
df_clusters["Cluster"] = kmeans_res

columns = ["BALANCE", "PURCHASES", "CASH_ADVANCE",
           "CREDIT_LIMIT", "PAYMENTS", "PURCHASES_FREQUENCY"]

cluster_profile = df_clusters.groupby("Cluster")[columns].mean()

print(cluster_profile.round(2))
```

```python
# Compare each cluster against overall average
scaled_df = pd.DataFrame(df_transform, columns=df.columns)
scaled_df["Cluster"] = kmeans_res

profile = scaled_df.groupby("Cluster").mean()

for cluster in profile.index:
    high = profile.loc[cluster].nlargest(2)
    low = profile.loc[cluster].nsmallest(2)

    print(f"\nCluster {cluster}:")
    print(f"Highest relative features: {list(high.index)}")
    print(f"Lowest relative features: {list(low.index)}")
```

### K-Means vs DBSCAN Comparison

K-Means divides credit card customers into four clusters based on their spending and payment behavior. DBSCAN identifies dense groups of customers and marks unusual observations as noise. K-Means assigns every customer to a cluster, whereas DBSCAN can leave some customers unassigned. The differences show how centroid-based and density-based clustering produce different customer segments.
